```python
import numpy as np
# Dari list Python
a1D = np.array([1, 2, 3, 4, 5]) # 1D
a2D = np.array([[1,2,3],[4,5,6]]) # 2D

# Array dengan nilai inisial
zeros = np.zeros((3, 4)) # nol semua
ones = np.ones((2, 3)) # satu semua
eye = np.eye(4) # matriks identitas
full = np.full((2, 2), 7) # nilai konstan

# Sequence
arange = np.arange(0, 20, 2) # 0,2,4,...18
linsp = np.linspace(0, 1, 11) # 11 titik merata

# Array acak (reproducible)
rng = np.random.default_rng(seed=42)
rand_int = rng.integers(0, 100, (3, 4))
rand_flt = rng.normal(loc=0, scale=1, size=(100,))
```

```python
# Operasi aritmetika vektorisasi
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])
print(a + b) # → [11 22 33 44 55]
print(a * b) # → [10 40 90 160 250]
print(a ** 2) # → [1 4 9 16 25]

# Universal functions (ufuncs)
np.sqrt(a) # akar kuadrat tiap elemen
np.exp(a) # eksponensial (e^x)
np.log(b) # logaritma natural

# Fungsi statistik
data = rng.normal(0, 1, 1000)
print(f'Mean: {np.mean(data):.4f}')
print(f'Std: {np.std(data):.4f}')
print(f'Min: {np.min(data):.4f}')
print(f'Max: {np.max(data):.4f}')
print(f'Q1: {np.percentile(data,25):.4f}')
print(f'Q3: {np.percentile(data,75):.4f}')
```

```python
# Membuat DataFrame dari dictionary
data = {
 'nama': ['Ali','Budi','Caca','Dina'],
 'usia': [22, 25, 21, 23],
 'kota': ['Jakarta','Bandung','Surabaya','Medan'],
 'ipk': [3.8, 3.5, 3.9, 3.7]
}
df = pd.DataFrame(data)

# Informasi dasar
print(df.shape) # → (4, 4)
print(df.dtypes) # tipe data per kolom
print(df.columns) # nama-nama kolom
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/refs/heads/master/titanic.csv"
df = pd.read_csv(url)
# Filter satu kondisi
penumpang_selamat = df[df['Survived'] == 1]
penumpang_wanita = df[df['Sex'] == 'female']

# Filter kondisi ganda (& = AND, | = OR)
wanita_kelas1 = df[(df['Sex']=='female') &
 (df['Pclass']==1)]

# Seleksi kolom tertentu
print(df)
print(df.columns)
```

```python
# Pola dasar: groupby → aggregation
df.groupby('Pclass')['Survived'].mean()

# Agregasi multiple kolom & fungsi
hasil = df.groupby('Pclass').agg(
 total = ('PassengerId', 'count'),
 selamat = ('Survived', 'sum'),
 pct_selamat= ('Survived', 'mean'),
 rata_usia = ('Age', 'mean'),
 max_fare = ('Fare', 'max')
)
hasil = hasil.round(3)
print(hasil)

# GroupBy multi-kolom
df.groupby(['Pclass','Sex'])['Survived']\
 .mean().unstack().round(3)
```

```python
missing = df.isnull().sum()
print(missing)
```

```python
missing_percent = (df.isnull().sum()/len(df))*100
print(missing_percent)
```

```python
missing_info = pd.DataFrame({
    'Missing Values' : df.isnull().sum(),
    'Percentage' : (df.isnull().sum()/len(df)) * 100
})

missing_info = missing_info[missing_info['Missing Values'] > 0]

print(missing_info)
```

```python
count = df['Survived'].value_counts()
percent = df['Survived'].value_counts(normalize=True) * 100

result = pd.DataFrame({
    'Jumlah' : count,
    'Presentase' : percent
}).reset_index()

result.columns = ['Status','Jumlah','Presentase']

result['Status'] = result['Status'].map({0: "Tidak Selamat", 1: "Selamat"})
print(result)
```

```python
# Filter kondisi ganda (& = AND, | = OR)
wanita_kelas1 = (df[(df['Sex']=='female') & (df['Pclass']==1)])
total = len(wanita_kelas1)
mean_age = wanita_kelas1['Age'].mean()

print(f'Passenger Wanita 1st Class : {total}')
print(f'Rata-rata Usia : {mean_age:.2f}')
```

```python
survived_byclass = df.groupby('Pclass')['Survived'].mean().sort_values(ascending=False)

print(survived_byclass)
```
