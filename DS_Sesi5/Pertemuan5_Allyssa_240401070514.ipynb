```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, matplotlib.gridspec as gridspec
import seaborn as sns

%matplotlib inline
```

```python
plt.close('all')

df = sns.load_dataset('diamonds')
print(df.shape, df.dtypes)
print(df.describe().round(2))
```

```python
fig = plt.figure(figsize=(16, 12))
fig.suptitle('Dashboard Analisis Dataset Diamonds',
             fontsize=18, fontweight='bold', y=0.98)
```

```python
gs = gridspec.GridSpec(2, 2, figure=fig,hspace=0.4, wspace=0.35)

ax1 = fig.add_subplot(gs[0, 0])
```

```python
sns.histplot(data=df, x='price', hue='clarity',
 kde=True, palette='Set2', ax=ax2)
ax2.set_title('Distribusi Harga per Clarity')
ax2.spines[['top','right']].set_visible(False)
```

```python
sns.scatterplot(
    data=df,
    x='carat',
    y='price',
    hue='cut',
    size='depth',
    sizes=(30, 180),
    palette='Set2',
    alpha=0.7,
    ax=ax4
)

sns.regplot(
    data=df,
    x='carat',
    y='price',
    scatter=False,
    color='gray',
    ax=ax4
)

ax4.set_title('Hubungan Carat & Price (Ukuran = Depth)')
ax4.set_xlabel('Carat')
ax4.set_ylabel('Price')
ax4.spines[['top','right']].set_visible(False)
```

```python
avg_by_clarity = df.groupby('clarity', observed=True)['price'].mean().sort_values()

ax1.bar(avg_by_clarity.index, avg_by_clarity.values,
        color='#028090', edgecolor='white')

ax1.set_title('Rata-rata Harga Setiap Clarity')
ax1.set_xlabel('Clarity')
ax1.set_ylabel('Harga')
ax1.tick_params(axis='x', rotation=0)
ax1.spines[['top','right']].set_visible(False)
```
